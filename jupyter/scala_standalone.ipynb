### To install new packages in the JupyterLab environment.
Use the following command:
import $ivy.`group::package:versión`

Example:
import $ivy.`org.typelevel::cats-core:2.9.0`
import cats._, cats.implicits._

```scala
import org.apache.spark.sql.SparkSession

val spark = SparkSession.builder()
  .master("spark://spark-cluster-master:7077")
  .appName("JupyterLabSparkScalaStandaloneMode")
  .config("spark.cores.max", "12")
  .config("spark.executor.cores", "3")
  .config("spark.executor.memory", "1g")
  .config("spark.driver.cores", "2")
  .getOrCreate()

println(s"Spark Version: ${spark.version}")
println(s"Spark context version: ${spark.sparkContext.version}")
println(s"Scala driver version: ${scala.util.Properties.versionNumberString}")
```

```scala
val data = Seq(("Alice", 29), ("Bob", 35), ("Cathy", 23))
val columns = Seq("Name", "Age")
val df = spark.createDataFrame(data).toDF(columns: _*)
df.show()
```

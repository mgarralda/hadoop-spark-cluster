```scala
// Import Spark and MLlib libraries
import org.apache.spark.sql.SparkSession
import org.apache.spark.ml.clustering.KMeans
import org.apache.spark.ml.feature.VectorAssembler
import org.apache.spark.sql.functions._

// Create Spark session
val spark = SparkSession.builder()
  .master("spark://spark-cluster-master:7077")
  .appName("JupyterLabSparkScalaMachineLearning")
  .config("spark.cores.max", "12")
  .config("spark.executor.cores", "3")
  .config("spark.executor.memory", "1g")
  .config("spark.driver.cores", "2")
  .getOrCreate()

import spark.implicits._

println(s"Spark Version: ${spark.version}")
```

```scala
// Sample DataFrame
val df = Seq(
  (0, 1.0, 2.0),
  (1, 1.5, 1.8),
  (2, 5.0, 8.0),
  (3, 8.0, 8.0),
  (4, 1.0, 0.6),
  (5, 9.0, 11.0)
).toDF("id", "x", "y")

// Assemble features into vector
val assembler = new VectorAssembler()
  .setInputCols(Array("x", "y"))
  .setOutputCol("features")

val assembled = assembler.transform(df)
```

```scala
// Train KMeans
val kmeans = new KMeans()
  .setK(2)
  .setSeed(1L)

val model = kmeans.fit(assembled)
```

```scala
// Predict cluster
val predictions = model.transform(assembled)
predictions.select("id", "features", "prediction").show(false)

// Show centers
println("Cluster Centers:")
model.clusterCenters.foreach(println)
```

```scala
spark.stop()
```

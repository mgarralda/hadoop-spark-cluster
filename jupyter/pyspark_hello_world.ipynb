### Checking the environment

```python
import sys
import os

print("🔍 Python path (sys.executable):")
print(sys.executable)

print("\n📦 PYSPARK_PYTHON env var:")
print(os.environ.get("PYSPARK_PYTHON"))

print("\n📦 PYSPARK_DRIVER_PYTHON env var:")
print(os.environ.get("PYSPARK_DRIVER_PYTHON"))

try:
    import pyspark
    print(f"\n✅ PySpark version: {pyspark.__version__}")

except ImportError as e:
    print("\n❌ pyspark or spark.ml not available:")
    print(e)
```
